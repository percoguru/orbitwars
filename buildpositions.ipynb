```python
import math

def euclidean_distance(p1: tuple[int, int], p2: tuple[int, int]) -> float:
    return math.sqrt((p2[0] - p1[0]) ** 2 + (p2[1] - p1[1]) ** 2)
```

```python
def future_positions(
    x: float,
    y: float,
    angular_speed: float,
    turns: int = 500,
    center: tuple[float, float] = (50.0, 50.0),
) -> list[tuple[float, float]]:
    cx, cy = center
    radius = math.sqrt((x - cx) ** 2 + (y - cy) ** 2)
    theta0 = math.atan2(y - cy, x - cx)

    return [
        (
            cx + radius * math.cos(theta0 + angular_speed * t),
            cy + radius * math.sin(theta0 + angular_speed * t),
        )
        for t in range(1, turns + 1)
    ]
```

```python
def turns_to_reach(
    start: tuple[float, float],
    end: tuple[float, float],
    speed: float,
) -> int:
    if path_crosses_sun(start, end):
        return -1
    return math.ceil(euclidean_distance(start, end) / speed)
```

```python
def path_crosses_sun(
    start: tuple[float, float],
    end: tuple[float, float],
    sun_center: tuple[float, float] = (50.0, 50.0),
    sun_radius: float = 10.0,
) -> bool:
    dx, dy = end[0] - start[0], end[1] - start[1]
    fx, fy = start[0] - sun_center[0], start[1] - sun_center[1]

    a = dx * dx + dy * dy
    b = 2 * (fx * dx + fy * dy)
    c = fx * fx + fy * fy - sun_radius ** 2

    # Degenerate segment: just check if the single point is inside the sun
    if a == 0:
        return c <= 0

    discriminant = b * b - 4 * a * c
    if discriminant < 0:
        return False

    disc_sqrt = math.sqrt(discriminant)
    t1 = (-b - disc_sqrt) / (2 * a)
    t2 = (-b + disc_sqrt) / (2 * a)

    # Segment intersects circle when [t1, t2] overlaps [0, 1]
    return t1 <= 1 and t2 >= 0
```
